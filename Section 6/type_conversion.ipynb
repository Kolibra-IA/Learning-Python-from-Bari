# TYPE CONVERSION

### Int Type Conversion

- It supports int, float, boolean, and string data types.

#### Int

```python
x = int(10)
print(x)
```

#### Float

```python
x = int(17.98)
print(x)
```

#### Boolean

```python
x = int(True)
print(x)
```

#### Complex

- Not supported!

#### String 

- Numerical strings are the valid arguments.

```python
x = int('1900')
print(x)
```

### Float Type Conversion

- It supports int, float, boolean, and valid strings.

#### Int

```python
x = float(10)
print(x)
```

#### Float

```python
x = float(10.99)
print(x)
```

#### Boolean

```python
x = float(True)
print(x)
```

#### Valid Strings

```python
x = float(10.99)
print(x)
```

### Boolean

- It supports int, float, boolean, complex and strings.

#### Int

```python
x = bool(12)
print(x)
```

#### Float

```python
x = bool(10.9)
print(x)
```

#### Boolean

```python
x = bool(False)
print(x)
```

#### Complex

```python
x = bool(10+5j)
print(x)
```

#### String

```python
x = bool('How una dey!')
print(x)
```

### Complex

- It supports int, float, boolean, complex and valid strings.

#### Int

```python
x = complex(29)
print(x)
```

#### Float 

```python
x = complex(10.98)
print(x)
```

#### Boolean

```python
x = complex(True)
print(x)
```

#### Complex

```python
x = complex(10+5j)
print(x)
```

#### String

```python
x = complex('10')
print(x)
```

### String

- It supports int, float, boolean, complex and string.

#### Int

```python
x = str(10)
print(x)
```

#### Float

```python
x = str(10.98)
print(x)
```

#### Boolean

```python
x = str(True)
print(x)
```

#### Complex 

```python
x = str(10+89j)
print(x)
```

#### String

```python
x = str('Up we go!')
print(x)
```
